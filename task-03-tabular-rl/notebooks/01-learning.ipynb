## 本 Notebook 的学习边界

环境模型未知时，价值通过样本而不是完整转移表学习。MC 使用完整回报：

$$
G_t=R_{t+1}+\gamma R_{t+2}+\gamma^2R_{t+3}+\cdots,
$$

TD(0) 使用一步 bootstrap：

$$
V(S_t)\leftarrow V(S_t)+lpha\left[R_{t+1}+\gamma V(S_{t+1})-V(S_t)ight].
$$

本 Notebook 的第一部分检查 prediction，第二部分比较 DP、MC、TD 以及 SARSA/Q-Learning；重点观察“是否 bootstrap”和“目标策略是否等于行为策略”。


````python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT))

from src.gridworld import GridWorld
from src.monte_carlo import generate_episode
from src.monte_carlo import first_visit_mc_prediction

import random

env = GridWorld()


def random_policy(state):
    return random.choice([0, 1, 2, 3])
````

````python
V_mc = first_visit_mc_prediction(
    env=env,
    policy=random_policy,
    num_episodes=100,
    gamma=1.0,
)

V_mc
````

````python
episode_counts = [
    10,
    100,
    1000,
    10000,
]

results = {}

for n in episode_counts:

    env = GridWorld()

    V = first_visit_mc_prediction(
        env=env,
        policy=random_policy,
        num_episodes=n,
        gamma=1.0,
    )

    results[n] = V

state = (2, 1)

for n in episode_counts:
    print(
        f"{n:5d} episodes: "
        f"V({state}) = {results[n][state]:.4f}"
    )
````

````python
from src.td_learning import td0_prediction

V = {
    (0, 0): -2.0,
    (0, 1): -5.0,
}

state = (0, 0)
next_state = (0, 1)

reward = -1.0

alpha = 0.1
gamma = 0.9

td_target = reward + gamma * V[next_state]

td_error = td_target - V[state]

V[state] += alpha * td_error

print("TD target:", td_target)
print("TD error:", td_error)
print("Updated V:", V[state])
````
