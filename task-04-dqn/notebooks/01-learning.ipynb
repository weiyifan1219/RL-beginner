## 本 Notebook 的学习边界

DQN 用神经网络近似动作价值：

$$
Q_	heta(s,a)pprox Q_*(s,a),\qquad y=r+\gamma(1-terminal)\max_{a'}Q_{	heta^-}(s',a').
$$
